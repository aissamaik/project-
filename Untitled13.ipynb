```python

import sqlite3

DB_PATH = "/content/miel_attendance.db"

connection = sqlite3.connect(DB_PATH)
connection.execute("PRAGMA foreign_keys = ON")

cursor = connection.cursor()

# 1. Modules
cursor.execute("""
CREATE TABLE IF NOT EXISTS modules (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name_fr TEXT NOT NULL,
    name_ar TEXT,
    semester TEXT NOT NULL,
    professor_id INTEGER,
    FOREIGN KEY(professor_id) REFERENCES users(id)
)
""")

# 2. Sessions
cursor.execute("""
CREATE TABLE IF NOT EXISTS sessions (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    module_id INTEGER NOT NULL,
    date TEXT NOT NULL,
    start_time TEXT NOT NULL,
    end_time TEXT,
    status TEXT DEFAULT 'scheduled',
    FOREIGN KEY(module_id) REFERENCES modules(id)
)
""")

# 3. Temporary attendance codes
cursor.execute("""
CREATE TABLE IF NOT EXISTS attendance_codes (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    session_id INTEGER NOT NULL,
    code_hash TEXT NOT NULL,
    expires_at TEXT NOT NULL,
    active INTEGER DEFAULT 1,
    FOREIGN KEY(session_id) REFERENCES sessions(id)
)
""")

# 4. Attendance records
cursor.execute("""
CREATE TABLE IF NOT EXISTS attendance (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    student_id INTEGER NOT NULL,
    session_id INTEGER NOT NULL,
    recorded_at TEXT DEFAULT CURRENT_TIMESTAMP,
    status TEXT DEFAULT 'present',
    UNIQUE(student_id, session_id),
    FOREIGN KEY(student_id) REFERENCES users(id),
    FOREIGN KEY(session_id) REFERENCES sessions(id)
)
""")

connection.commit()

print("All tables created successfully!")

connection.close()
```

```python

import sqlite3
from werkzeug.security import generate_password_hash

DB_PATH = "/content/miel_attendance.db"

def register_student(full_name, email, password):

    full_name = full_name.strip()
    email = email.strip().lower()

    if not full_name or not email or not password:
        return "All fields are required"

    if len(password) < 8:
        return "Password must contain at least 8 characters"

    password_hash = generate_password_hash(password)

    connection = sqlite3.connect(DB_PATH)
    cursor = connection.cursor()

    try:
        cursor.execute("""
        INSERT INTO users (
            full_name,
            email,
            password_hash,
            role,
            status
        )
        VALUES (?, ?, ?, 'student', 'pending')
        """, (full_name, email, password_hash))

        connection.commit()

        return "Registration successful! Awaiting approval."

    except sqlite3.IntegrityError:
        return "This email is already registered."

    finally:
        connection.close()
```

```python

import sqlite3
from werkzeug.security import check_password_hash

DB_PATH = "/content/miel_attendance.db"

def login_student(email, password):

    email = email.strip().lower()

    connection = sqlite3.connect(DB_PATH)
    cursor = connection.cursor()

    cursor.execute("""
    SELECT id, full_name, password_hash, status
    FROM users
    WHERE email = ? AND role = 'student'
    """, (email,))

    user = cursor.fetchone()
    connection.close()

    if user is None:
        return "Invalid email or password"

    user_id, full_name, password_hash, status = user

    if not check_password_hash(password_hash, password):
        return "Invalid email or password"

    if status != "approved":
        return "Your account is awaiting approval"

    return {
        "id": user_id,
        "name": full_name,
        "status": status
    }
```

```python

from flask import (
    Flask, request, redirect,
    url_for, session, flash,
    render_template_string
)
import sqlite3
import os
from werkzeug.security import (
    generate_password_hash,
    check_password_hash
)

app = Flask(__name__)
app.secret_key = os.urandom(32)

DB_PATH = "/content/miel_attendance.db"

# Design
STYLE = """
<style>
body {
    font-family: Arial, sans-serif;
    background: #f1f5f9;
    margin: 0;
    padding: 30px 15px;
    text-align: center;
}
.card {
    background: white;
    max-width: 400px;
    margin: 50px auto;
    padding: 30px;
    border-radius: 15px;
    box-shadow: 0 4px 20px #0001;
}
h1 { color: #17365d; }
input {
    width: 90%;
    padding: 12px;
    margin: 10px 0;
    border: 1px solid #ddd;
    border-radius: 7px;
}
button {
    background: #17365d;
    color: white;
    padding: 12px 25px;
    border: none;
    border-radius: 7px;
    cursor: pointer;
}
a { color: #2563eb; }
.message { color: #d97706; }
</style>
"""

# Register page
@app.route("/register", methods=["GET", "POST"])
def register():

    if request.method == "POST":

        name = request.form.get("name", "").strip()
        email = request.form.get("email", "").strip().lower()
        password = request.form.get("password", "")

        if not name or not email or not password:
            return "All fields are required"

        if len(password) < 8:
            return "Password must contain at least 8 characters"

        password_hash = generate_password_hash(password)

        connection = sqlite3.connect(DB_PATH)

        try:
            connection.execute("""
            INSERT INTO users
            (full_name, email, password_hash, role, status)
            VALUES (?, ?, ?, 'student', 'pending')
            """, (name, email, password_hash))

            connection.commit()

            return """
            <h2>Registration successful!</h2>
            <p>Your account is awaiting approval.</p>
            <a href="/login">Login</a>
            """

        except sqlite3.IntegrityError:
            return "This email is already registered."

        finally:
            connection.close()

    return render_template_string(STYLE + """
    <div class="card">
        <h1>MIEL Attendance</h1>
        <h3>Créer un compte étudiant</h3>

        <form method="POST">
            <input name="name" placeholder="Nom complet" required>
            <input name="email" type="email"
                   placeholder="Email" required>
            <input name="password" type="password"
                   placeholder="Mot de passe" required>
            <button type="submit">S'inscrire</button>
        </form>

        <p>Déjà inscrit ? <a href="/login">Connexion</a></p>
    </div>
    """)


# Login page
@app.route("/login", methods=["GET", "POST"])
def login():

    if request.method == "POST":

        email = request.form.get("email", "").strip().lower()
        password = request.form.get("password", "")

        connection = sqlite3.connect(DB_PATH)

        user = connection.execute("""
        SELECT id, full_name, password_hash, status, role
        FROM users
        WHERE email = ?
        """, (email,)).fetchone()

        connection.close()

        if user is None:
            return "Invalid email or password"

        if not check_password_hash(user[2], password):
            return "Invalid email or password"

        if user[3] != "approved":
            return "Your account is awaiting approval"

        session["user_id"] = user[0]
        session["name"] = user[1]
        session["role"] = user[4]

        return redirect(url_for("dashboard"))

    return render_template_string(STYLE + """
    <div class="card">
        <h1>MIEL Attendance</h1>
        <h3>Connexion</h3>

        <form method="POST">
            <input name="email" type="email"
                   placeholder="Email" required>
            <input name="password" type="password"
                   placeholder="Mot de passe" required>
            <button type="submit">Se connecter</button>
        </form>

        <p>Pas de compte ? <a href="/register">Inscription</a></p>
    </div>
    """)


# Student dashboard
@app.route("/dashboard")
def dashboard():

    if "user_id" not in session:
        return redirect(url_for("login"))

    return render_template_string(STYLE + """
    <div class="card">
        <h1>MIEL Attendance</h1>
        <h3>Bienvenue {{ name }}</h3>

        <p>Votre espace étudiant est actif.</p>
        <p>Votre rôle : {{ role }}</p>

        <p>Gestion de présence bientôt disponible.</p>

        <a href="/logout">Déconnexion</a>
    </div>
    """, name=session["name"], role=session["role"])


# Logout
@app.route("/logout")
def logout():

    session.clear()

    return redirect(url_for("login"))


print("MIEL Attendance Flask App created successfully!")
```

```python

import sqlite3
import secrets
import hashlib
from datetime import datetime, timedelta, timezone

def generate_attendance_code(session_id, professor_id):

    connection = sqlite3.connect(DB_PATH)
    cursor = connection.cursor()

    # Verify that the session belongs to this professor
    session_info = cursor.execute("""
        SELECT s.id
        FROM sessions s
        JOIN modules m ON s.module_id = m.id
        WHERE s.id = ? AND m.professor_id = ?
    """, (session_id, professor_id)).fetchone()

    if not session_info:
        connection.close()
        return "Session not found or access denied"

    # Generate a random 6-digit code
    code = f"{secrets.randbelow(1000000):06d}"

    # Store only the hash, not the original code
    code_hash = hashlib.sha256(code.encode()).hexdigest()

    # Expiration time: 5 minutes
    expires_at = (
        datetime.now(timezone.utc) + timedelta(minutes=5)
    ).isoformat()

    # Deactivate previous codes for this session
    cursor.execute("""
        UPDATE attendance_codes
        SET active = 0
        WHERE session_id = ?
    """, (session_id,))

    cursor.execute("""
        INSERT INTO attendance_codes
        (session_id, code_hash, expires_at, active)
        VALUES (?, ?, ?, 1)
    """, (session_id, code_hash, expires_at))

    connection.commit()
    connection.close()

    # Display original code only to the professor
    return {
        "code": code,
        "expires_at": expires_at
    }

print("Attendance code generator created successfully!")
```

```python

import sqlite3
import hashlib
from datetime import datetime, timezone

def verify_attendance_code(student_id, code):

    connection = sqlite3.connect(DB_PATH)
    cursor = connection.cursor()

    student = cursor.execute("""
        SELECT id, status
        FROM users
        WHERE id = ? AND role = 'student'
    """, (student_id,)).fetchone()

    if not student or student[1] != "approved":
        connection.close()
        return "Student account not approved"

    code_hash = hashlib.sha256(code.strip().encode()).hexdigest()

    record = cursor.execute("""
        SELECT id, session_id, expires_at
        FROM attendance_codes
        WHERE code_hash = ? AND active = 1
    """, (code_hash,)).fetchone()

    if not record:
        connection.close()
        return "Invalid or inactive code"

    code_id, session_id, expires_at = record

    expiry = datetime.fromisoformat(expires_at)

    if datetime.now(timezone.utc) >= expiry:
        cursor.execute("""
            UPDATE attendance_codes
            SET active = 0
            WHERE id = ?
        """, (code_id,))
        connection.commit()
        connection.close()
        return "Code expired"

    try:
        cursor.execute("""
            INSERT INTO attendance (
                student_id, session_id, recorded_at, status
            )
            VALUES (?, ?, ?, 'present')
        """, (
            student_id,
            session_id,
            datetime.now(timezone.utc).isoformat()
        ))

        connection.commit()
        result = "Attendance recorded successfully!"

    except sqlite3.IntegrityError:
        result = "Attendance already recorded"

    connection.close()
    return result


print("Attendance verification function created!")
```

```python
def professor_enrollments():

    if "user_id" not in session:
        return redirect("/login")

    if session.get("role") != "professor":
        return "Access denied", 403

    connection = sqlite3.connect(DB_PATH)
    cursor = connection.cursor()

    message = None

    if request.method == "POST":
        module_id = request.form.get("module_id", type=int)
        student_id = request.form.get("student_id", type=int)

        owned_module = cursor.execute("""
            SELECT id FROM modules
            WHERE id = ? AND professor_id = ?
        """, (module_id, session["user_id"])).fetchone()

        approved_student = cursor.execute("""
            SELECT id FROM users
            WHERE id = ? AND role = 'student'
            AND status = 'approved'
        """, (student_id,)).fetchone()

        if not owned_module:
            message = "Module invalide ou accès refusé."

        elif not approved_student:
            message = "Étudiant invalide ou non approuvé."

        else:
            try:
                cursor.execute("""
                    INSERT INTO enrollments (student_id, module_id)
                    VALUES (?, ?)
                """, (student_id, module_id))

                connection.commit()
                message = "Inscription ajoutée avec succès !"

            except sqlite3.IntegrityError:
                message = "Cet étudiant est déjà inscrit."

    modules = cursor.execute("""
        SELECT id, name_fr
        FROM modules
        WHERE professor_id = ?
    """, (session["user_id"],)).fetchall()

    students = cursor.execute("""
        SELECT id, full_name, email
        FROM users
        WHERE role = 'student' AND status = 'approved'
        ORDER BY full_name
    """).fetchall()

    enrollments = cursor.execute("""
        SELECT u.full_name, m.name_fr
        FROM enrollments e
        JOIN users u ON u.id = e.student_id
        JOIN modules m ON m.id = e.module_id
        WHERE m.professor_id = ?
        ORDER BY m.name_fr, u.full_name
    """, (session["user_id"],)).fetchall()

    connection.close()

    return render_template_string(
        ENROLLMENTS_PAGE,
        modules=modules,
        students=students,
        enrollments=enrollments,
        message=message
    )


def dashboard_router():
    view = request.args.get("view")

    if view == "modules":
        return professor_modules()
    elif view == "sessions":
        return professor_sessions()
    elif view == "attendance":
        return student_attendance()
    elif view == "professor-code":
        return professor_code_page()
    elif view == "professor-attendance":
        return professor_attendance_page()
    elif view == "export-excel":
        return professor_export_excel()
    elif view == "enrollments":
        return professor_enrollments()

    return premium_dashboard()


app.view_functions["dashboard"] = dashboard_router

print("Enrollment management connected successfully!")
```
